# 实验 2：复现 6 位镜像对称实验

本 Notebook 复现 Rumelhart、Hinton 与 Williams（1986）论文 *Learning representations by back-propagating errors* 的第一个原始实验：训练一个 `6 -> 2 -> 1` 的 Sigmoid 网络，判断 6 位二进制向量是否关于中心镜像对称。

## 本实验要回答的问题

1. 按论文给出的网络、损失、全批量更新、学习率、动量、初始化范围和训练轮数，能否正确分类全部 64 个输入？
2. 两个隐藏单元是否学到论文图 1 所示的结构：
   - 对称位置权重近似等幅反号；
   - 两个隐藏单元的输入权重近似互为相反数；
   - 三组权重绝对值近似 `1:2:4`；
   - 隐藏到输出的权重为负，输出偏置为正？
3. 去掉隐藏层后，线性模型能否完成任务？
4. 原论文未报告随机种子时，结果对初始化有多敏感？

## 验收标准

- 手写反向传播再次通过全部 17 个参数的有限差分检查；
- 代表性运行在 64 个输入上达到 `64/64`；
- 最小正例概率大于最大负例概率；
- 成功解的配对反对称误差 `< 1e-3`、隐藏权重余弦相似度 `< -0.999`；
- 排序后的三组权重幅值接近 `1:2:4`；
- 无隐藏层对照不能达到 `64/64`。


## 1. 论文实验协议

| 项目 | 论文设置 | 本 Notebook |
|---|---:|---:|
| 输入 | 6 位二进制向量 | 相同，穷举全部 64 个 |
| 正例条件 | $x_1=x_6,\ x_2=x_5,\ x_3=x_4$ | 相同 |
| 网络 | `6 -> 2 -> 1` | 相同 |
| 激活 | Sigmoid | 相同 |
| 损失 | 半平方误差总和 | 相同，不除以 batch 大小 |
| 更新 | 每遍历 64 个样本后更新 | 相同，全批量梯度和 |
| 初始化 | $U(-0.3,0.3)$ | 相同，权重和偏置均初始化 |
| 学习率 | $\varepsilon=0.1$ | 相同 |
| 动量 | $\alpha=0.9$ | 相同 |
| 训练长度 | 1,425 sweeps | 相同，不提前停止 |

论文没有公开随机种子和伪随机数生成器。本 Notebook 将种子 `1` 固定为代表性成功运行，并额外报告种子 `0..99` 的完整结果分布。这里的目标是复现论文的**机制与现象**，不是得到逐位相同的权重。


```python
# 环境、依赖与实验配置
from __future__ import annotations

from collections import Counter
from itertools import product
import platform
import sys
from time import perf_counter

import matplotlib.pyplot as plt
import numpy as np

CONFIG = {
    "primary_seed": 1,
    "epochs": 1_425,
    "learning_rate": 0.1,
    "momentum": 0.9,
    "init_scale": 0.3,
    "classification_threshold": 0.5,
    "finite_difference_h": 1e-5,
    "seed_trials": 100,
}

np.set_printoptions(precision=6, suppress=True)
plt.style.use("seaborn-v0_8-whitegrid")

environment = {
    "python": sys.version.split()[0],
    "numpy": np.__version__,
    "platform": platform.platform(),
}
print("Configuration:", CONFIG)
print("Environment:", environment)
```

## 2. 实验计划

1. 穷举 64 个输入并验证其中恰有 8 个镜像对称模式；
2. 在本 Notebook 中重新给出自包含的 NumPy MLP；
3. 用全部 64 条数据检查 17 个参数的解析梯度；
4. 固定种子 `1`，严格训练 1,425 sweeps；
5. 检查分类、概率间隔和论文描述的隐藏权重结构；
6. 训练无隐藏层的 `6 -> 1` 对照；
7. 重复 100 个种子，区分成功、全负塌缩和部分成功三类结果。


## 3. 构造全部 64 个输入

一个向量镜像对称，当且仅当它具有 `abc|cba` 的形式。因此自由位只有 3 个，正例数量应为 $2^3=8$；其余 56 个都是负例。


```python
X = np.array(list(product([0.0, 1.0], repeat=6)), dtype=np.float64)
y = np.all(X == X[:, ::-1], axis=1).astype(np.float64)[:, None]
patterns = np.array(["".join(str(int(bit)) for bit in row) for row in X])

assert X.shape == (64, 6)
assert y.shape == (64, 1)
assert X.dtype == np.float64 and y.dtype == np.float64
assert int(y.sum()) == 8
assert np.all(y[:, 0].astype(bool) == (
    (X[:, 0] == X[:, 5])
    & (X[:, 1] == X[:, 4])
    & (X[:, 2] == X[:, 3])
))

positive_patterns = patterns[y[:, 0] == 1]
print(f"Dataset shape: {X.shape}")
print(f"Symmetric / non-symmetric: {int(y.sum())} / {len(y) - int(y.sum())}")
print("All symmetric patterns:", ", ".join(positive_patterns))
```

## 4. 自包含的基础 MLP

与 `1_MLP.ipynb` 一致，但这里不导入外部 `.py` 文件，保证 Notebook 可独立运行。

对每层：

$$
Z^{(l)}=A^{(l-1)}W^{(l)}+b^{(l)},\qquad
A^{(l)}=\sigma(Z^{(l)}).
$$

损失严格采用论文的 batch 求和：

$$
E=\frac12\sum_{c,j}(\hat y_{c,j}-y_{c,j})^2.
$$

动量更新为

$$
\Delta\theta_t=-\varepsilon\nabla_\theta E_t+\alpha\Delta\theta_{t-1},
\qquad \theta_t=\theta_{t-1}+\Delta\theta_t.
$$


```python
def sigmoid(x: np.ndarray) -> np.ndarray:
    '''Numerically stable sigmoid in float64.'''
    x = np.asarray(x, dtype=np.float64)
    output = np.empty_like(x)
    positive = x >= 0
    output[positive] = 1.0 / (1.0 + np.exp(-x[positive]))
    exp_x = np.exp(x[~positive])
    output[~positive] = exp_x / (1.0 + exp_x)
    return output


def sigmoid_grad_from_output(activation: np.ndarray) -> np.ndarray:
    return activation * (1.0 - activation)


class MLP:
    '''Fully connected sigmoid MLP with manual backpropagation.'''

    def __init__(
        self,
        layer_sizes: list[int] | tuple[int, ...],
        *,
        seed: int,
        init_scale: float,
    ) -> None:
        if len(layer_sizes) < 2:
            raise ValueError("layer_sizes must include input and output widths")
        if any(not isinstance(width, (int, np.integer)) or width <= 0
               for width in layer_sizes):
            raise ValueError("all layer widths must be positive integers")
        if init_scale <= 0:
            raise ValueError("init_scale must be positive")

        self.layer_sizes = tuple(int(width) for width in layer_sizes)
        rng = np.random.default_rng(seed)
        self.params: dict[str, np.ndarray] = {}
        self.grads: dict[str, np.ndarray] = {}
        self.velocity: dict[str, np.ndarray] = {}
        self.cache: list[dict[str, np.ndarray]] = []

        for layer, (n_in, n_out) in enumerate(
            zip(self.layer_sizes[:-1], self.layer_sizes[1:]), start=1
        ):
            self.params[f"W{layer}"] = rng.uniform(
                -init_scale, init_scale, size=(n_in, n_out)
            ).astype(np.float64)
            self.params[f"b{layer}"] = rng.uniform(
                -init_scale, init_scale, size=(n_out,)
            ).astype(np.float64)

        for name, parameter in self.params.items():
            self.grads[name] = np.zeros_like(parameter)
            self.velocity[name] = np.zeros_like(parameter)

    @property
    def n_layers(self) -> int:
        return len(self.layer_sizes) - 1

    @property
    def n_parameters(self) -> int:
        return sum(parameter.size for parameter in self.params.values())

    def forward(self, X_batch: np.ndarray, *, store_cache: bool = True) -> np.ndarray:
        activation = np.asarray(X_batch, dtype=np.float64)
        if activation.ndim != 2 or activation.shape[1] != self.layer_sizes[0]:
            raise ValueError(
                f"X must have shape (batch, {self.layer_sizes[0]}), "
                f"got {activation.shape}"
            )

        cache: list[dict[str, np.ndarray]] = []
        for layer in range(1, self.n_layers + 1):
            W = self.params[f"W{layer}"]
            b = self.params[f"b{layer}"]
            z = activation @ W + b
            next_activation = sigmoid(z)
            cache.append({"a_prev": activation, "z": z, "a": next_activation})
            activation = next_activation

        if store_cache:
            self.cache = cache
        return activation

    @staticmethod
    def squared_error(prediction: np.ndarray, target: np.ndarray) -> float:
        return float(0.5 * np.sum((prediction - target) ** 2))

    def loss(self, X_batch: np.ndarray, y_batch: np.ndarray) -> float:
        prediction = self.forward(X_batch, store_cache=False)
        target = np.asarray(y_batch, dtype=np.float64)
        if target.shape != prediction.shape:
            raise ValueError(f"target shape {target.shape} != {prediction.shape}")
        return self.squared_error(prediction, target)

    def loss_and_backward(
        self, X_batch: np.ndarray, y_batch: np.ndarray
    ) -> tuple[float, np.ndarray]:
        prediction = self.forward(X_batch, store_cache=True)
        target = np.asarray(y_batch, dtype=np.float64)
        if target.shape != prediction.shape:
            raise ValueError(f"target shape {target.shape} != {prediction.shape}")

        loss = self.squared_error(prediction, target)
        d_activation = prediction - target

        for layer in range(self.n_layers, 0, -1):
            layer_cache = self.cache[layer - 1]
            a_prev = layer_cache["a_prev"]
            activation = layer_cache["a"]
            delta = d_activation * sigmoid_grad_from_output(activation)

            self.grads[f"W{layer}"] = a_prev.T @ delta
            self.grads[f"b{layer}"] = delta.sum(axis=0)
            d_activation = delta @ self.params[f"W{layer}"].T

        return loss, prediction

    def step(self, learning_rate: float, *, momentum: float) -> None:
        if learning_rate <= 0:
            raise ValueError("learning_rate must be positive")
        if not 0.0 <= momentum < 1.0:
            raise ValueError("momentum must satisfy 0 <= momentum < 1")

        for name in self.params:
            self.velocity[name] = (
                momentum * self.velocity[name]
                - learning_rate * self.grads[name]
            )
            self.params[name] += self.velocity[name]
```

## 5. 在当前任务上重新做梯度检查

`6 -> 2 -> 1` 网络共有

$$
6\times2+2+2\times1+1=17
$$

个参数。下面在全部 64 条数据上逐元素比较解析梯度和中心有限差分梯度。


```python
def gradient_check(
    model: MLP,
    X_batch: np.ndarray,
    y_batch: np.ndarray,
    *,
    h: float,
) -> list[dict[str, float | int | str]]:
    reference_loss, _ = model.loss_and_backward(X_batch, y_batch)
    analytical = {name: gradient.copy() for name, gradient in model.grads.items()}
    report: list[dict[str, float | int | str]] = []

    for name, parameter in model.params.items():
        numerical = np.zeros_like(parameter)
        for index in np.ndindex(parameter.shape):
            original = parameter[index]
            parameter[index] = original + h
            loss_plus = model.loss(X_batch, y_batch)
            parameter[index] = original - h
            loss_minus = model.loss(X_batch, y_batch)
            parameter[index] = original
            numerical[index] = (loss_plus - loss_minus) / (2.0 * h)

        absolute_error = np.abs(analytical[name] - numerical)
        relative_error = absolute_error / np.maximum(
            1e-12, np.abs(analytical[name]) + np.abs(numerical)
        )
        report.append({
            "parameter": name,
            "elements": parameter.size,
            "max_absolute_error": float(absolute_error.max()),
            "max_relative_error": float(relative_error.max()),
        })

    assert np.isclose(model.loss(X_batch, y_batch), reference_loss)
    return report


check_model = MLP(
    [6, 2, 1], seed=CONFIG["primary_seed"], init_scale=CONFIG["init_scale"]
)
gradient_report = gradient_check(
    check_model, X, y, h=CONFIG["finite_difference_h"]
)

print(f"{'parameter':<10} {'n':>4} {'max abs error':>18} {'max rel error':>18}")
print("-" * 56)
for row in gradient_report:
    print(
        f"{row['parameter']:<10} {row['elements']:>4d} "
        f"{row['max_absolute_error']:>18.3e} "
        f"{row['max_relative_error']:>18.3e}"
    )

gradient_max_abs = max(row["max_absolute_error"] for row in gradient_report)
gradient_max_rel = max(row["max_relative_error"] for row in gradient_report)

assert check_model.n_parameters == 17
assert sum(int(row["elements"]) for row in gradient_report) == 17
assert gradient_max_abs < 1e-8
assert gradient_max_rel < 1e-6
print("Exhaustive 17-parameter gradient check passed.")
```

## 6. 训练与评估函数

`train_model` 每个 sweep 都在全部 64 个样本上计算**梯度和**，随后更新一次参数；没有 mini-batch、数据打乱或提前停止。


```python
def train_model(
    layer_sizes: list[int] | tuple[int, ...],
    *,
    seed: int,
    record_history: bool = True,
) -> tuple[MLP, np.ndarray | None]:
    model = MLP(
        layer_sizes,
        seed=seed,
        init_scale=CONFIG["init_scale"],
    )
    history = (
        np.empty(CONFIG["epochs"], dtype=np.float64)
        if record_history else None
    )

    for sweep in range(CONFIG["epochs"]):
        loss, _ = model.loss_and_backward(X, y)
        if history is not None:
            history[sweep] = loss
        model.step(
            CONFIG["learning_rate"],
            momentum=CONFIG["momentum"],
        )
    return model, history


def binary_metrics(
    model: MLP,
    *,
    threshold: float = CONFIG["classification_threshold"],
) -> dict[str, float | int | np.ndarray]:
    probabilities = model.forward(X, store_cache=False).ravel()
    predictions = (probabilities >= threshold).astype(int)
    targets = y.ravel().astype(int)

    tp = int(np.sum((predictions == 1) & (targets == 1)))
    tn = int(np.sum((predictions == 0) & (targets == 0)))
    fp = int(np.sum((predictions == 1) & (targets == 0)))
    fn = int(np.sum((predictions == 0) & (targets == 1)))
    tpr = tp / (tp + fn)
    tnr = tn / (tn + fp)

    return {
        "probabilities": probabilities,
        "predictions": predictions,
        "loss": model.loss(X, y),
        "accuracy": (tp + tn) / len(y),
        "balanced_accuracy": 0.5 * (tpr + tnr),
        "tp": tp,
        "tn": tn,
        "fp": fp,
        "fn": fn,
        "positive_min": float(probabilities[targets == 1].min()),
        "negative_max": float(probabilities[targets == 0].max()),
    }
```

## 7. 代表性论文协议运行

固定种子 `1`，严格执行 1,425 次全批量更新。种子只决定 $U(-0.3,0.3)$ 中抽到的初始值，不改变任何论文超参数。


```python
started = perf_counter()
symmetry_model, symmetry_history = train_model(
    [6, 2, 1], seed=CONFIG["primary_seed"], record_history=True
)
primary_runtime = perf_counter() - started
primary = binary_metrics(symmetry_model)

assert symmetry_history is not None
assert primary["tp"] == 8 and primary["tn"] == 56
assert primary["fp"] == 0 and primary["fn"] == 0
assert primary["accuracy"] == 1.0
assert primary["positive_min"] > primary["negative_max"]

print(f"Final loss:            {primary['loss']:.10f}")
print(f"Accuracy:              {int(64 * primary['accuracy'])}/64")
print(f"Confusion (TP/TN/FP/FN): {primary['tp']}/{primary['tn']}/{primary['fp']}/{primary['fn']}")
print(f"Minimum positive prob: {primary['positive_min']:.6f}")
print(f"Maximum negative prob: {primary['negative_max']:.6f}")
print(f"Probability margin:    {primary['positive_min'] - primary['negative_max']:.6f}")
print(f"Runtime:               {primary_runtime:.3f} s")
```

```python
# 展示全部 8 个正例，以及最容易被误判为正例的 8 个负例。
probabilities = primary["probabilities"]
positive_indices = np.flatnonzero(y[:, 0] == 1)
negative_indices = np.flatnonzero(y[:, 0] == 0)
hardest_negative_indices = negative_indices[
    np.argsort(probabilities[negative_indices])[-8:][::-1]
]

print("Symmetric patterns")
print(f"{'pattern':<10} {'target':>7} {'probability':>13} {'class':>7}")
print("-" * 41)
for index in positive_indices:
    predicted = int(probabilities[index] >= CONFIG["classification_threshold"])
    print(f"{patterns[index]:<10} {int(y[index, 0]):>7d} "
          f"{probabilities[index]:>13.6f} {predicted:>7d}")

print("\nHardest non-symmetric patterns")
print(f"{'pattern':<10} {'target':>7} {'probability':>13} {'class':>7}")
print("-" * 41)
for index in hardest_negative_indices:
    predicted = int(probabilities[index] >= CONFIG["classification_threshold"])
    print(f"{patterns[index]:<10} {int(y[index, 0]):>7d} "
          f"{probabilities[index]:>13.6f} {predicted:>7d}")
```

## 8. 分析隐藏层学到的算法

如果一个隐藏单元对每组镜像位置学到等幅反号权重，则其预激活近似为

$$
z_h\approx a_1(x_1-x_6)+a_2(x_2-x_5)+a_3(x_3-x_4)+b_h.
$$

当 $|a_1|,|a_2|,|a_3|$ 排序后约为 `1:2:4` 时，任何非零差分组合都不会完全抵消。两个隐藏单元再使用近乎相反的权重，分别检测加权差为正和为负的情况：

- 对称输入：三个差都为 0，两个隐藏单元都被负偏置压低；
- 非对称输入：至少一个差非零，至少一个隐藏单元被激活；
- 输出层：两个隐藏到输出的权重均为负，只有“两者都低”时，正输出偏置才能产生高输出。


```python
W1 = symmetry_model.params["W1"]
b1 = symmetry_model.params["b1"]
W2 = symmetry_model.params["W2"].ravel()
b2 = float(symmetry_model.params["b2"][0])

# Rows correspond to (x1,x6), (x2,x5), (x3,x4).
pair_sums = W1[:3, :] + W1[::-1][:3, :]
pair_relative_errors = (
    np.linalg.norm(pair_sums, axis=0)
    / np.linalg.norm(W1[:3, :], axis=0)
)
hidden_cosine = float(
    np.dot(W1[:, 0], W1[:, 1])
    / (np.linalg.norm(W1[:, 0]) * np.linalg.norm(W1[:, 1]))
)
hidden_opposite_relative_error = float(
    np.linalg.norm(W1[:, 0] + W1[:, 1]) / np.linalg.norm(W1[:, 0])
)

magnitude_ratios = np.sort(np.abs(W1[:3, :]).T, axis=1)
magnitude_ratios /= magnitude_ratios[:, :1]

hidden_activations = sigmoid(X @ W1 + b1)
symmetric_hidden_max = float(hidden_activations[y[:, 0] == 1].max())
nonsymmetric_detector_min = float(
    np.max(hidden_activations[y[:, 0] == 0], axis=1).min()
)

print("Learned input -> hidden weights (rows x1..x6, columns h1/h2):")
print(W1)
print("\nHidden biases:", b1)
print("Hidden -> output weights:", W2)
print("Output bias:", f"{b2:.6f}")
print("\nMirror-pair weight sums (ideal = 0):")
print(pair_sums)
print("Pair antisymmetry relative errors:", pair_relative_errors)
print("Cosine similarity between hidden weight vectors:", f"{hidden_cosine:.10f}")
print("Hidden-opposite relative error:", f"{hidden_opposite_relative_error:.6f}")
print("Sorted magnitude ratios by hidden unit:")
print(magnitude_ratios)
print("Max hidden activation on symmetric cases:", f"{symmetric_hidden_max:.6f}")
print("Min max-hidden activation on non-symmetric cases:",
      f"{nonsymmetric_detector_min:.6f}")

assert np.all(pair_relative_errors < 1e-3)
assert hidden_cosine < -0.999
assert hidden_opposite_relative_error < 0.01
assert np.allclose(
    magnitude_ratios,
    np.array([[1.0, 2.0, 4.0], [1.0, 2.0, 4.0]]),
    atol=0.08,
    rtol=0.0,
)
assert np.all(b1 < 0.0)
assert np.all(W2 < 0.0)
assert b2 > 0.0
assert symmetric_hidden_max < 0.2
assert nonsymmetric_detector_min > 0.5
print("\nAll structural reproduction checks passed.")
```

## 9. 无隐藏层对照

`6 -> 1` Sigmoid 模型在阈值 `0.5` 下仍然只是线性分类器。使用相同初始化范围、学习率、动量和 1,425 sweeps，检查它能否完成任务。

由于正例只有 8 个，单看普通准确率会产生误导：永远预测负类也有 $56/64=87.5\%$。因此同时报告正例召回率所参与的 balanced accuracy 和混淆计数。


```python
linear_model, linear_history = train_model(
    [6, 1], seed=CONFIG["primary_seed"], record_history=True
)
linear = binary_metrics(linear_model)

assert linear_history is not None
assert linear["accuracy"] < 1.0
assert linear["tp"] < 8

print(f"{'model':<18} {'params':>7} {'loss':>12} {'accuracy':>11} "
      f"{'balanced':>11} {'TP/TN/FP/FN':>15}")
print("-" * 82)
print(f"{'6 -> 2 -> 1':<18} {symmetry_model.n_parameters:>7d} "
      f"{primary['loss']:>12.6f} {primary['accuracy']:>10.2%} "
      f"{primary['balanced_accuracy']:>10.2%} "
      f"{primary['tp']}/{primary['tn']}/{primary['fp']}/{primary['fn']:>2}")
print(f"{'6 -> 1':<18} {linear_model.n_parameters:>7d} "
      f"{linear['loss']:>12.6f} {linear['accuracy']:>10.2%} "
      f"{linear['balanced_accuracy']:>10.2%} "
      f"{linear['tp']}/{linear['tn']}/{linear['fp']}/{linear['fn']:>2}")
print("\nThe no-hidden-layer control does not solve all 64 patterns.")
```

## 11. 100 个随机种子的复现稳定性

论文未公开随机种子，并明确讨论了局部极小值。这个任务又有明显类别不平衡：若网络把所有输出压到 0，半平方误差约为

$$
\frac12\times 8=4,
$$

且表面准确率仍有 87.5%。因此对种子 `0..99` 严格重复同一协议，并将运行分类为：

- `64/64`：全部正确；
- `all-negative`：8 个正例全部漏掉；
- `partial`：其余情况。


```python
seed_results: list[dict[str, float | int | str]] = []
started = perf_counter()

for seed in range(CONFIG["seed_trials"]):
    trial_model, _ = train_model([6, 2, 1], seed=seed, record_history=False)
    trial = binary_metrics(trial_model)

    if trial["accuracy"] == 1.0:
        state = "64/64"
    elif trial["tp"] == 0 and trial["fp"] == 0:
        state = "all-negative"
    else:
        state = "partial"

    seed_results.append({
        "seed": seed,
        "loss": float(trial["loss"]),
        "accuracy": float(trial["accuracy"]),
        "tp": int(trial["tp"]),
        "tn": int(trial["tn"]),
        "fp": int(trial["fp"]),
        "fn": int(trial["fn"]),
        "state": state,
    })

seed_sweep_runtime = perf_counter() - started
state_counts = Counter(row["state"] for row in seed_results)
successful_seed_ids = [row["seed"] for row in seed_results if row["state"] == "64/64"]

assert CONFIG["primary_seed"] in successful_seed_ids
assert state_counts["64/64"] > 0
assert sum(state_counts.values()) == CONFIG["seed_trials"]

print("Outcome counts:")
for state in ["64/64", "partial", "all-negative"]:
    print(f"  {state:<13}: {state_counts[state]:>3d}")
print(f"Success rate: {state_counts['64/64']}/{CONFIG['seed_trials']} "
      f"({state_counts['64/64'] / CONFIG['seed_trials']:.1%})")
print("Successful seed IDs:", successful_seed_ids)
print(f"Seed sweep runtime: {seed_sweep_runtime:.3f} s")
```

```python
state_order = ["64/64", "partial", "all-negative"]
state_colors = {
    "64/64": "tab:green",
    "partial": "tab:orange",
    "all-negative": "tab:red",
}

fig, axes = plt.subplots(1, 2, figsize=(12, 4.3))

for state in state_order:
    rows = [row for row in seed_results if row["state"] == state]
    if rows:
        axes[0].scatter(
            [row["seed"] for row in rows],
            [row["loss"] for row in rows],
            s=34, alpha=0.8, color=state_colors[state], label=state
        )
axes[0].axhline(4.0, color="gray", linestyle="--", linewidth=1.2)
axes[0].set_yscale("log")
axes[0].set(
    xlabel="Seed", ylabel="Final loss", title="Initialization sensitivity"
)
axes[0].legend()

counts = [state_counts[state] for state in state_order]
axes[1].bar(
    state_order, counts,
    color=[state_colors[state] for state in state_order], alpha=0.85
)
for index, count in enumerate(counts):
    axes[1].text(index, count + 1, str(count), ha="center", va="bottom")
axes[1].set(
    xlabel="Outcome", ylabel="Number of seeds",
    title=f"Outcomes across {CONFIG['seed_trials']} seeds",
    ylim=(0, max(counts) * 1.15)
)

fig.tight_layout()
plt.show()
```

## 13. 结论与复现边界

### 已复现的核心现象

1. 固定种子 `1` 严格执行论文协议后，网络正确分类全部 `64/64` 个模式；
2. 隐藏层权重对镜像位置形成近乎精确的等幅反号结构；
3. 两个隐藏权重向量近乎互为相反数，三组幅值接近 `1:2:4`；
4. 对称输入使两个隐藏单元都保持低激活，非对称输入至少激活一个隐藏单元；
5. 两个负的隐藏到输出权重配合正输出偏置，实现“两个隐藏单元都低才判为对称”；
6. 无隐藏层模型不能完成任务，说明隐藏非线性是必要组件。

### 额外发现

本次 100 种子实验得到 `25/100` 个 `64/64` 解；74 个种子塌缩为全负预测，另 1 个部分成功。损失约为 4 的全负解正对应 8 个正例被全部漏掉。这说明：**在严格保持论文超参数、且网络只有两个隐藏单元时，初始化对能否离开类别不平衡平台非常重要。**

### 不能声称的内容

- 论文没有给出随机种子、数值精度和完整初始化序列，因此无法逐权重复现图 1；
- 本次权重绝对值与图 1 不同，但符号关系、反对称结构、`1:2:4` 比例和实现功能一致；
- 训练使用现代 NumPy `float64`，不能等同于 1986 年的具体软硬件实现。
